# Exploring the Languages of the Top 500 Novels

For this assignment, I chose the **Top 500 “Greatest” Novels** dataset from [Responsible Datasets in Context](https://www.responsible-datasets-in-context.com/posts/top-500-novels/top-500-novels.html). It combines a library-based ranking with information about novels and their authors.

**Question:** Which original languages appear most often in this list?

I use Pandas to read the CSV file directly from the dataset project on GitHub.

```python
import pandas as pd
import altair as alt

url = "https://raw.githubusercontent.com/melaniewalsh/responsible-datasets-in-context/refs/heads/main/datasets/top-500-novels/top-500-novels-metadata_2025-01-11.csv"
df = pd.read_csv(url)
df.head()
```

Make an Altair chart, count how many novels have each original language and plot the ten most common. 

```python
language_counts = (
    df["orig_lang"]
    .fillna("Unknown")
    .value_counts()
    .head(10)
    .rename_axis("Original language")
    .reset_index(name="Number of novels")
)

chart = (
    alt.Chart(language_counts)
    .mark_bar()
    .encode(
        x=alt.X("Number of novels:Q", title="Number of novels"),
        y=alt.Y("Original language:N", sort="-x", title="Original language"),
        tooltip=[
            alt.Tooltip("Original language:N"),
            alt.Tooltip("Number of novels:Q"),
        ],
    )
    .properties(title="Most common original languages in the Top 500 novels", width=550, height=320)
    .interactive()
)
chart
```

What I noticed

English is much more common than other original languages in this list. The chart makes the difference in representation easy to see. This is interesting because the ranking is based on books held by libraries, not on every novel ever published.

The dataset authors explain that the ranking depends on the libraries represented in WorldCat, which are not equally distributed around the world. This chart describes this particular list, not which languages produce the most literature or the best novels.
